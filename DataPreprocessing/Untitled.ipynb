```python
currPath=Path.cwd()
normal_path = currPath.parents[1]/"Datasets/DESeq2Input/lungNormalCounts.csv"
cancer_path = currPath.parents[1]/"Datasets/DESeq2Input/lungCancerCounts.csv"
output_folder_normal = currPath.parents[1]/"Datasets/DESeq2Input/LungNormalSingle/"
output_folder_cancer = currPath.parents[1]/"Datasets/DESeq2Input/LungCancerSingle/"
```

```python
# Read the CSV, assuming the first column has gene identifiers and 
# the rest of the columns are samples with counts.
df = pd.read_csv(cancer_path, index_col=0)

# For each sample (i.e., each column in df), export a separate CSV.
for sample_name in df.columns:
    # Extract just this sample’s counts, keeping the genes as rows.
    sample_df = df[[sample_name]]
    
    # Write out to a new CSV named after the sample
    output_file = f"{output_folder_cancer}/cancer_{sample_name}.csv"
    sample_df.to_csv(output_file)
    
    print(f"Saved: {output_file}")
```

```python
# Read the CSV, assuming the first column has gene identifiers and 
# the rest of the columns are samples with counts.
df = pd.read_csv(normal_path, index_col=0)

# For each sample (i.e., each column in df), export a separate CSV.
for sample_name in df.columns:
    # Extract just this sample’s counts, keeping the genes as rows.
    sample_df = df[[sample_name]]
    
    # Write out to a new CSV named after the sample
    output_file = f"{output_folder_normal}/normal_{sample_name}.csv"
    sample_df.to_csv(output_file)
    
    print(f"Saved: {output_file}")
```
